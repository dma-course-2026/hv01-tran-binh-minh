```python
#Bài 53
def traugamco():
    sotrau=100
    soco=100
    trauco=[]
    for traudung in range (0,sotrau):
        for traungoi in range (0,sotrau):
            for traugia in range (0,sotrau):
                if traudung+traungoi+traugia==sotrau and traudung*5+traungoi*3+traugia/3==soco:
                    trauco.append((traudung,traungoi,traugia))
    return trauco
print (traugamco())

    
```

```python
#Bài 57
def fibonacci():
    n=10
    fibonac=[]
    a,b=0,1
    for i in range(n):
        fibonac.append(a)
        a,b=b,a+b
    print(f"{n}số Fibonacci đầu tiên là :{fibonac}")
    return fibonac
fibonacci() 

```

```python
#Bài 58
def uoc_lon_va_nho():
    a=int(input("Nhập số nguyên dương a:"))
    b=int(input("Nhập số nguyên dương b:"))
    USCLN=1
    for i in range(1,min(a,b) +1):
        if a%i==0 and b%i==0:
            BSNN=a*b//USCLN
            print(f"uscln của {a}và{b}là: {USCLN}")
            print(f"bsnn của {a}và {b}là:{BSNN}")
uoc_lon_va_nho()
```

```python
#Bài 59
def lietke():
    n = int(input("Nhập số nguyên dương n: "))
    result=[]
    for a in range(2,n*n+2):
        uoc=0
        for b in range(2,int(a**0.5)+1):
            if a%b==0:
               uoc+=1
            if uoc==0 and len(result)<n:
                result.append(a)
    print (f"{n}số nguyên đầu tiên là : {result}")
    return result
lietke()
```

```python
#Bài 61
def liet_ke_tat_ca():
    so_nguyen_to_5_chu_so=[]
    for a in range (10000,100000):
        uoc=0
        for b in range (2,int(a**0.5)+1):
            if a%b==0:
                uoc+=1
        if uoc==0:
            so_nguyen_to_5_chu_so.append(a)
            print(f"Các số nguyên tố có 5 chữ số là: {so_nguyen_to_5_chu_so}")
            print(f"Số lượng các số nguyên tố có 5 chữ số là: {len(so_nguyen_to_5_chu_so)}")
            return so_nguyen_to_5_chu_so
liet_ke_tat_ca()

```
