```python
#Bài 98
def xaukitu():
    input_string = input("Nhập vào một xâu: ")
    result_string = ''.join(sorted(set(input_string), key=input_string.index))
    print(f"Xâu sau khi loại bỏ các kí tự trùng lặp: {result_string}")
    return result_string
xaukitu()
```

```python
#Bài 99
def tachso():
    input_string = input("Nhập vào một xâu: ")
    result_string = ''.join(filter(lambda x: not x.isdigit(), input_string))
    print(f"Xâu sau khi loại bỏ các số: {result_string}")
    return result_string
tachso()
```

```python
#Bài 100
def tach_xau():
    input_string = input("Nhập vào một xâu: ")
    result_string = ''.join(filter(lambda x: not x.isalpha(), input_string))
    print(f"Xâu sau khi loại bỏ các chữ cái: {result_string}")
    return result_string
tach_xau()
```
