```python
#Bài 2
class GiaiThua:
    def tinh(self, n):
        if n == 0 or n == 1:
            return 1
        ket_qua = 1
        for i in range(1, n + 1):
            ket_qua *= i
        return ket_qua
n = int(input("Nhập số n: "))
finder = GiaiThua()
print(str(finder.tinh(n)))
    
```

```python
#Bài 3
class dictionary:
    def __init__(self, n):
        self.n = n
    def tao(self):
        return {i: i * i for i in range(1, self.n + 1)}
finder = dictionary(8)
print(finder.tao())
```

```python
#Bài 9
from datetime import date
class TinhTuoi:
    def __init__(self, ngay_sinh_str):
        nam, thang, ngay = map(int, ngay_sinh_str.split('/'))
        self.ngay_sinh = date(nam, thang, ngay)
    def tinh(self):
        today = date.today()
        tuoi = today.year - self.ngay_sinh.year  
        return tuoi
chuoi_ngay_sinh = input("Nhập ngày tháng năm sinh (dạng y/m/d): ")
nguoi = TinhTuoi(chuoi_ngay_sinh)
print(f"Số tuổi của bạn là: {nguoi.tinh()}")
```

```python
#Bài 10
class TinhLuong:
    def __init__(self, so_gio, luong_gio):
        self.so_gio = so_gio
        self.luong_gio = luong_gio
    def tinh_thuc_linh(self):
        gio_tieu_chuan = 44
        if self.so_gio <= gio_tieu_chuan:
            thuc_linh = self.so_gio * self.luong_gio
        else:
            gio_lam_them = self.so_gio - gio_tieu_chuan
            thuc_linh = (gio_tieu_chuan * self.luong_gio) + (gio_lam_them * self.luong_gio * 1.5)
        return thuc_linh
gio = float(input("Nhập số giờ làm mỗi tuần: "))
luong = float(input("Nhập thù lao mỗi giờ tiêu chuẩn: "))
nv = TinhLuong(gio, luong)
print(f"Số tiền thực lĩnh của nhân viên là: {nv.tinh_thuc_linh()}")
```

```python
#Bài 11
import math
class TinhQ:
    def __init__(self, c=50, h=30):
        self.C = c  
        self.H = h
    def tinh_chuoi_d(self, d_str):
        danh_sach_d = d_str.split(',')
        ket_qua = []
        for d in danh_sach_d:
            q = round(math.sqrt((2 * self.C * float(d)) / self.H))
            ket_qua.append(str(q))
        return ",".join(ket_qua)
chuoi_input = input("Nhập dãy giá trị D (phân tách bởi dấu phẩy): ")
calculator = TinhQ()
print(calculator.tinh_chuoi_d(chuoi_input))
```

```python
#Bài 12
class MangHaiChieu:
    def __init__(self, x, y):
        self.X = x
        self.Y = y
    def tao_mang(self):
        mang_2d = []
        for i in range(self.X):
            hang = []
            for j in range(self.Y):
                hang.append(i * j)
            mang_2d.append(hang)
        return mang_2d
input_str = input("Nhập giá trị X, Y (phân tách bởi dấu phẩy): ")
x, y = map(int, input_str.split(','))
generator = MangHaiChieu(x, y)
print(generator.tao_mang())
```

```python
#Bài 13
class SapXepTu:
    def __init__(self, chuoi_input):
        self.chuoi_input = chuoi_input
    def sap_xep(self):
        danh_sach_tu = [tu.strip() for tu in self.chuoi_input.split(',')]
        danh_sach_tu.sort()
        return ",".join(danh_sach_tu)
chuoi_nhap = input("Nhập các từ phân tách bởi dấu phẩy: ")
sorter = SapXepTu(chuoi_nhap)
print(sorter.sap_xep())
```

```python
#Bài 15
class XuLyChuoi:
    def __init__(self, chuoi_input):
        self.chuoi_input = chuoi_input
    def xu_ly(self):
        danh_sach_tu = self.chuoi_input.split()
        tap_tu_duy_nhat = set(danh_sach_tu)
        danh_sach_sap_xep = sorted(tap_tu_duy_nhat)
        return " ".join(danh_sach_sap_xep)
chuoi_nhap = input("Nhập chuỗi các từ: ")
processor = XuLyChuoi(chuoi_nhap)
print(processor.xu_ly())
```

```python
#Bài 16
class KiemTraNhiPhan:
    def __init__(self, chuoi_input):
        self.chuoi_input = chuoi_input
    def tim_so_chia_het_cho_5(self):
        danh_sach_nhi_phan = [p.strip() for p in self.chuoi_input.split(',')]
        ket_qua = []
        for p in danh_sach_nhi_phan:
            gia_tri_thap_phan = int(p, 2)
            if gia_tri_thap_phan % 5 == 0:
                ket_qua.append(p)
        return ",".join(ket_qua)
chuoi_nhap = input("Nhập các số nhị phân 4 chữ số (phân tách bởi dấu phẩy): ")
checker = KiemTraNhiPhan(chuoi_nhap)
print(checker.tim_so_chia_het_cho_5())
```

```python
#Bài 17
class TimSoChan:
    def __init__(self, bat_dau=1000, ket_thuc=3000):
        self.bat_dau = bat_dau
        self.ket_thuc = ket_thuc
    def tim_kiem(self):
        ket_qua = []
        for i in range(self.bat_dau, self.ket_thuc + 1):
            s = str(i)
            if (int(s[0]) % 2 == 0) and (int(s[1]) % 2 == 0) and (int(s[2]) % 2 == 0) and (int(s[3]) % 2 == 0):
                ket_qua.append(s)
        return ",".join(ket_qua)
finder = TimSoChan(1000, 3000)
print(finder.tim_kiem())
```

```python
#Bài 18
class DemChuVaSo:
    def __init__(self, cau):
        self.cau = cau
        self.so_chu_cai = 0
        self.so_chu_so = 0
    def dem(self):
        for c in self.cau:
            if c.isalpha():
                self.so_chu_cai += 1
            elif c.isdigit():
                self.so_chu_so += 1
        return self.so_chu_cai, self.so_chu_so
cau_nhap = input("Nhập vào một câu: ")
dem_obj = DemChuVaSo(cau_nhap)
chu, so = dem_obj.dem()
print(f"Số chữ cái là: {chu}")
print(f"Số chữ số là: {so}")
```

```python
#Bài 19
class DemHoaThuong:
    def __init__(self, cau):
        self.cau = cau
        self.so_chu_hoa = 0
        self.so_chu_thuong = 0
    def dem(self):
        for c in self.cau:
            if c.isupper():
                self.so_chu_hoa += 1
            elif c.islower():
                self.so_chu_thuong += 1
        return self.so_chu_hoa, self.so_chu_thuong
cau_nhap = input("Nhập vào một câu: ")
dem_obj = DemHoaThuong(cau_nhap)
hoa, thuong = dem_obj.dem()
print(f"Chữ hoa: {hoa}")
print(f"Chữ thường: {thuong}")
```

```python
#Bài 21
class LocSoLe:
    def __init__(self, chuoi_input):
        self.chuoi_input = chuoi_input
    def loc(self):
        danh_sach_so = [int(x) for x in self.chuoi_input.split(',')]
        so_le = [str(x) for x in danh_sach_so if x % 2 != 0]
        return ",".join(so_le)
chuoi_nhap = input("Nhập danh sách các số (phân tách bởi dấu phẩy): ")
filter_obj = LocSoLe(chuoi_nhap)
print(filter_obj.loc())
```

```python
#Bài 22
class TaiKhoanNganHang:
    def __init__(self):
        self.so_du = 0
    def xu_ly_nhat_ky(self):
        print("Nhập nhật ký giao dịch (nhấn Enter trên dòng trống để kết thúc):")
        while True:
            dong = input()
            if not dong:
                break
            cac_phan = dong.split()
            loai_giao_dich = cac_phan[0].upper()
            so_tien = int(cac_phan[1])
            if loai_giao_dich == "Deposit":
                self.so_du += so_tien  
            elif loai_giao_dich == "Withdrawal":
                self.so_du -= so_tien  
        return self.so_du
account = TaiKhoanNganHang()
print(account.xu_ly_nhat_ky())
```

```python
#Bài 23
class DayDoiXung:
    def __init__(self, lst):
        self.lst = list(lst)
    def kiem_tra_doi_xung(self):
        return self.lst == self.lst[::-1]
    def sx_doi_xung(self):
        if self.kiem_tra_doi_xung():
            return self.lst
        n = len(self.lst)
        for i in range(n // 2):
            self.lst[n - 1 - i] = self.lst[i]
        return self.lst
input_str = input("Nhập dãy số nguyên (phân tách bởi dấu phẩy): ")
danh_sach = [int(x) for x in input_str.split(',')]
dx_obj = DayDoiXung(danh_sach)
if dx_obj.kiem_tra_doi_xung():
    print("Dãy số đã cho là dãy đối xứng.")
else:
    print("Dãy số chưa đối xứng (khả đối ứng).")
    day_moi = dx_obj.sx_doi_xung()
    print("Dãy số sau khi biến đổi thành đối xứng:", day_moi)
```
