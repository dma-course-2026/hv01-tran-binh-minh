```python
#Bài 20
def phuong_trinh_bac_2():
    a = float(input("Nhap he so a: "))
    b = float(input("Nhap he so b: "))
    c = float(input("Nhap he so c: "))
    delta = b ** 2 - 4 * a * c
    if delta < 0:
        print("Phuong trinh vo nghiem")
    elif delta == 0:
        x = -b / (2 * a)
        print(f"Phuong trinh co nghiem kep: {x}")
    else:
        x1 = (-b + delta ** 0.5) / (2 * a)
        x2 = (-b - delta ** 0.5) / (2 * a)
        print(f"Phuong trinh co hai nghiem phan biet: x1 = {x1}, x2 = {x2}")
    return delta
phuong_trinh_bac_2()
```

```python
#Bài 22
def tinh_chan_le():
    a = int(input("Nhap so nguyen thu nhat: "))
    b = int(input("Nhap so nguyen thu hai: "))
    if a % 2 == 0 and b % 2 == 0:
        print(a, "va", b, "la hai so cung chan")
    elif a % 2 != 0 and b % 2 != 0:
        print(a, "va", b, "la hai so cung le")
    else:
        print(a, "va", b, "khong cung chan cung le")
    return a, b
tinh_chan_le()
```

```python
#Bài 25
def gia_tri_lon_nhat_3_so():
    a = int(input("Nhap so nguyen thu nhat: "))
    b = int(input("Nhap so nguyen thu hai: "))
    c = int(input("Nhap so nguyen thu ba: "))
    if a > b and a > c:
        print(a, "la gia tri lon nhat")
    elif b > a and b > c:
        print(b, "la gia tri lon nhat")
    else:
        print(c, "la gia tri lon nhat")
    return a, b, c
gia_tri_lon_nhat_3_so()

```

```python
#Bài 28
def tinh_2_ngoi():
    a = int(input("Nhap so nguyen thu 1: "))
    b = int(input("Nhap so nguyen thu 2: "))
    pheptoan = input("Nhap cho phep toan: ")
    if pheptoan == "+":
        ketqua = a + b
        print(f"Tổng hai số là = {ketqua}")
    elif pheptoan == "-":
        ketqua = a - b
        print(f"Hiệu hai số là = {ketqua}")
    elif pheptoan == "*":
        ketqua = a * b
        print(f"Tích hai số là = {ketqua}")
    elif pheptoan == "/":
        if b != 0:
            ketqua = a / b
            print(f"Thương hai số là = {ketqua:.2f}")
        else:
            print("Khong the chia cho 0")
    else:
        print("Phep toan khong hop le")
    return a, b, pheptoan
tinh_2_ngoi()
```
