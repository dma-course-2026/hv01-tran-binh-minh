```python
#Bài 20
def phuong_trinh_bac_2():
    a = float(input("Nhap he so a: "))
    b = float(input("Nhap he so b: "))
    c = float(input("Nhap he so c: "))
    delta = b ** 2 - 4 * a * c
    if delta < 0:
        print("Phuong trinh vo nghiem")
    elif delta == 0:
        x = -b / (2 * a)
        print(f"Phuong trinh co nghiem kep: {x}")
    else:
        x1 = (-b + delta ** 0.5) / (2 * a)
        x2 = (-b - delta ** 0.5) / (2 * a)
        print(f"Phuong trinh co hai nghiem phan biet: x1 = {x1}, x2 = {x2}")
    return delta
phuong_trinh_bac_2()
```

```python
#Bài 32
def hephuongtrinhbac1haian():
    a, b, m, c, d, n = map(int, input("Nhap a, b, m, c, d, n: ").split(","))
    D = a * d - b * c
    Dx = m * d - b * n
    Dy = a * n - c * m
    if D == 0 and Dx == Dy == 0:
        print ("Phương trình có vô số nghiệm")
    elif D == 0 and Dx > 0 and Dy > 0:
        print ("phương trình vô nghiệm")
    else:
        x = Dx/D
        y = Dy/D
        print (f"Hệ phương trình có 1 cặp nghiệm là {x} và {y}")
    return x, y
hephuongtrinhbac1haian()
```
