```python
#Bài 13
import math

# Nhập bán kính R
r = float(input("Nhập bán kính R: "))

# Tính diện tích (S = π * R²) và chu vi (C = 2 * π * R)
dien_tich = math.pi * (r ** 2)
chu_vi = 2 * math.pi * r

# In kết quả làm tròn 2 chữ số thập phân
print(f"Diện tích {dien_tich:.2f}, Chu vi {chu_vi:.2f}")
```

```python
#Bài 14
import math

# Nhập 3 cạnh a, b, c trên cùng 1 dòng
a, b, c = map(float, input("Nhập 3 cạnh a, b, c: ").split())

# Kiểm tra điều kiện tạo thành tam giác
if a + b > c and a + c > b and b + c > a:
    # Tính chu vi
    chu_vi = a + b + c
    
    # Tính nửa chu vi (p) để dùng công thức Heron
    p = chu_vi / 2
    
    # Tính diện tích bằng công thức Heron
    dien_tich = math.sqrt(p * (p - a) * (p - b) * (p - c))
    
    # In kết quả làm tròn 2 chữ số thập phân
    print(f"Chu vi {chu_vi:.2f}, Diện tích {dien_tich:.2f}")
else:
    print("Ba số nhập vào không tạo thành tam giác.")
```

```python
#Bài 20
import math
a, b, c = map(float, input("Nhập hệ số a, b, c: ").split())

if a == 0:
    if b == 0:
        if c == 0:
            print("Phương trình có vô số nghiệm .")
        else:
            print("Phương trình vô nghiệm .")
    else:
        x = -c / b
        print(f"x = {x:.1f}")
else:
    delta = b**2 - 4 * a * c
    
    if delta < 0:
        print("Phương trình vô nghiệm .")
    elif delta == 0:
        x = -b / (2 * a)
        print(f"Phương trình có nghiệm kép: x = {x:.1f}")
    else:
        x1 = (-b + math.sqrt(delta)) / (2 * a)
        x2 = (-b - math.sqrt(delta)) / (2 * a)
        print(f"x1 = {x1:.1f}, x2 = {x2:.1f} .")
```

```python
#Bài 33 
import math
x0, y0, r = map(float, input("Nhập x0, y0 và R: ").split())
xa, ya = map(float, input("Nhập xa, ya của điểm A: ").split())
d = math.sqrt((xa - x0)**2 + (ya - y0)**2)
if abs(d - r) < 1e-9:
    print("Điểm A nằm trên đường tròn .")
elif d < r:
    print("Điểm A nằm trong đường tròn .")
else:
    print("Điểm A nằm ngoài đường tròn .")
```
