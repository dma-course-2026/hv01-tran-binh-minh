```python
#Bài 2
a= int(input("Nhập số a: "))
b= int(input("Nhập số b: "))
tong = a+b
hieu = a-b
tich = a*b
thuong = a/b
print(f"Tổng của {a} và {b} là {tong}")
print(f"Hiệu của {a} và {b} là {hieu}")
print(f"Tích của {a} và {b} là {tich}")
print(f"Thương của {a} và {b} là {thuong}")
```
